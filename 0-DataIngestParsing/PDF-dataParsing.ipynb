### here  in this file we are gonna learn how to extract the data from a pdf file and mainly there are two ways-->

1. Using PyPDFLoader
    - simple and reliable
    - used for only text pdf's(no special icopns and images)
    - keeps the copunt of pages
2. using PyMuPDFLoader
    - also simple
    - no page count
    - can also extract images
    - Fast extract

```python
# using PyPDFLoader ---> pdf from the data/pdf folder
from langchain_community.document_loaders import PyPDFLoader

pdf_loader = PyPDFLoader("data/pdf/Subhadeep_Deb_Resume.pdf")
pdf = pdf_loader.load()

print(pdf)
print(f"Content of the book are : {pdf[0].page_content}")

```

```python
# using PyMuPDFLoader ---> pdf from the data/pdf folder

## much fastewr compared to the pypdf
from langchain_community.document_loaders import PyMuPDFLoader

pdf_mu_loader = PyMuPDFLoader("data/pdf/Subhadeep_Deb_Resume.pdf")
pdf_ = pdf_mu_loader.load()

print(pdf_)
print(f"Content of the book are : {pdf_[0].page_content}")
```
