#### This file is made to understand text ingestion!!
I mean how datas are extracted from a text file and converted langchain document structure

```python
#creating a data directory inside the current directory
import os
os.makedirs('data/text_files', exist_ok = True) # creating a data folder and inside that another folder named "text_files" which will contain only text files
```

```python
# creating a dictionary where i will use keys as the path for files and values as their contents to create text files. We can do this using another methids also but lets use this only!!

all_files = {
    "data/text_files/first.txt": """Lorem Ipsum is simply dummy text of the printing and typesetting industry. Lorem Ipsum has been the industry's standard dummy text ever since 1966, when designers at Letraset and James Mosley, the librarian at St Bride Printing Library in London, took a 1914 Cicero translation and scrambled it to make dummy text for Letraset's Body Type sheets. It has survived not only many decades, but also the leap into electronic typesetting, remaining essentially unchanged. It was popularised thanks to these sheets and more recently with desktop publishing software like Aldus PageMaker and Microsoft Word including versions of Lorem Ipsum.""", # used """ for multiline content

    "data/text_files/second.txt":"""It is a long established fact that a reader will be distracted by the readable content of a page when looking at its layout. The point of using Lorem Ipsum is that it has a more-or-less normal distribution of letters, as opposed to using 'Content here, content here', making it look like readable English. Many desktop publishing packages and web page editors now use Lorem Ipsum as their default model text, and a search for 'lorem ipsum' will uncover many web sites still in their infancy. Various versions have evolved over the years, sometimes by accident, sometimes on purpose (injected humour and the like)."""
}

for path,content in all_files.items(): # using items() on dictionaries we can iterate on their values
    with open(path,'w',encoding = 'utf-8') as f:
        f.write(content)
```

```python
# read from a single txt file---> TextLoader
import langchain_community
from langchain_community.document_loaders import TextLoader

loader = TextLoader("data/text_files/first.txt", encoding='utf-8')
print(type(loader)) #---> TextLoader type
print(loader)

list_of_documents = loader.load()
print(list_of_documents)
print(type(list_of_documents)) #--> list of langchain Documents(contains page_content and metadata)

print(f"The length of characters in the Doc : {len(list_of_documents[0].page_content)}")
print(list_of_documents[0].metadata)
```

```python
#extracting data from all text files a directory ---> Directoryloader
from langchain_community.document_loaders import DirectoryLoader

loader = DirectoryLoader(  # ----> all the files unside the directory have to be of same type means if text then only text files
    "data/text_files",
    glob = "**/*.txt",
    loader_cls= TextLoader, #--> using which loader class those each file will get load
    loader_kwargs = { "encoding" : "utf-8"},
    show_progress= True
)

print(type(loader)) #--> DirectoryLoader

list_of_docs = loader.load()
print(type(list_of_docs)) #--> list of langchain docs

for i,doc in enumerate(list_of_docs):
    print(f"The file number {i+1}: ")
    print(f"The length of characters in the Doc : {len(list_of_docs[i].page_content)}")
    print(list_of_docs[i].metadata)
    print()# --> keeps a empty line 
```
