```python
# CREATE A JSONL FILE ( JSONL MEANS JSON LINE BY LINE)

jsonl_data=[
    {"employee_id":1,"name":"John","department":"Engineering","skills":["Python","Docker","Kubernetes"],"project":"RAG Chatbot","status":"Active"},
    {"employee_id":2,"name":"Emma","department":"Engineering","skills":["Java","Spring Boot"],"project":"Payment Gateway","status":"Completed"},
    {"employee_id":3,"name":"Sophia","department":"Data Science","skills":["PyTorch","LangChain"],"project":"LLM Research","status":"Active"},
    {"employee_id":4,"name":"David","department":"HR","skills":["Recruitment","Payroll"],"project":"Hiring Drive","status":"Planning"},
    {"employee_id":5,"name":"Lisa","department":"Marketing","skills":["SEO","Analytics"],"project":"Product Launch","status":"Completed"},
]

with open("data/json_files/company.jsonl",'w') as f:
    for item in jsonl_data:
        f.write(json.dumps(item) + "\n")  # json.dumps--> convert it to string as write expects a string argument
```

```python
from langchain_community.document_loaders import JSONLoader

dept_loader = JSONLoader(
    file_path = "data/json_files/company.json",
    jq_schema= ".company.departments[]",
    text_content= False
)

dept_loader_docs = dept_loader.load()
print(dept_loader_docs)
```
