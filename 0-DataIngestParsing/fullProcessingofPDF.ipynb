so here i will make a class which can process a whole pdf from loading thecontents to chunking

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from typing import List
from langchain_core.documents import Document
from langchain_community.document_loaders import PyPDFLoader


class Pdfprocessor:
    def __init__(self,chunk_size=20,chunk_overlap=5):
        self.chunkSize = chunk_size
        self.chunkOverlap = chunk_overlap
        self.text_splitter = RecursiveCharacterTextSplitter(
            separators = [" "],
            chunk_size = self.chunkSize,
            chunk_overlap = self.chunkOverlap
        )
    
    def clear_text(self,text:str)->str:
        final_txt = " ".join(text.split())
        return final_txt

    def process(self,path:str)-> List[Document]:
        processed = PyPDFLoader(path)
        pages = processed.load()

        # now we will do the chunking
        processed_chunks_full = []
        for page_num, page in enumerate(pages):
            cleared_text = self.clear_text(page.page_content)

            chunks = self.text_splitter.create_documents(  # creates list of documenets of each chunk of below format
                texts = [cleared_text],
                metadatas = [{
                    **page.metadata,
                    "page": page_num + 1
                }]
            )

            processed_chunks_full.extend(chunks)

        return processed_chunks_full

```

```python
pdfprocesser = Pdfprocessor(30,10)
```

```python
data = pdfprocesser.process("data/pdf/Subhadeep_Deb_Resume.pdf")
print(data)
```
