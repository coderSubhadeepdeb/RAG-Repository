This is created to describe about `text splitters` only. 
Why we need to split the texts that we got from a file content
- The the `LLM` models have a context size limit so we have to split the text or content into chunks
- We can do mainly 3 types of splitting 
    - character text splitting
    - recursive character text splitting
    - token based text splitting


1. **CharacterTextSplitter**: splitting based on text and used when you know only separator is there
2. **REcursiveCharacterTextSplitter**: splitting based on text and mostly used for text chunking as taking  as deafult splitter(in humans mind)
3. **TokenTextSplitter**: splitting based on token size and mostly used for models who are token limited. This splitter technique is a bit slower

```python
# importing all the text splitters
from langchain_text_splitters import (
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter,
    TokenTextSplitter
)
```

```python
# Getting the text to split
from langchain_community.document_loaders import TextLoader

loader = TextLoader("data/text_files/first.txt", encoding='utf-8')
list_of_documents = loader.load()
text = list_of_documents[0].page_content
```

```python
# Using the CharacterTextSplitter

# create a oblect of the CharacterTextSplitter class
chartxtsplitt = CharacterTextSplitter(
    separator= " ", #---> the "CharacterTextSplitter" provides only one way to split the text
    chunk_size = 20,
    chunk_overlap = 5,
    length_function = len
)

all_chunks = chartxtsplitt.split_text(text)
print(len(all_chunks))
print(all_chunks) #---> list of all chunks
```

```python
# Using the RecursiveCharacterTextSplitter

# create a oblect of the CharacterTextSplitter class
recursivechartxtsplitt = RecursiveCharacterTextSplitter(
    separators= [" ", "", "\n"], #---> the "RecursiveCharacterTextSplitter" provides multiple way to split the text
    chunk_size = 20,
    chunk_overlap = 5, # later helps to get all the chunks related to each other at the time of data retrieval from DB cause there is an overlap which helps to get similar related chunks in similar DB search
    length_function = len
)

all_recur_chunks = recursivechartxtsplitt.split_text(text)
print(len(all_recur_chunks))
print(all_recur_chunks) #---> list of all chunks
```

```python
# Using the TokenTextSplitter

#creating an object
tokentxtsplitt = TokenTextSplitter(
    chunk_size = 20, #--> size in tokens not characters
    chunk_overlap = 5, # later helps to get all the chunks related to each other at the time of data retrieval from DB cause there is an overlap which helps to get similar related chunks in similar DB search
)

all_token_chunks = tokentxtsplitt.split_text(text)
print(len(all_recur_chunks))
print(all_recur_chunks) #---> list of all chunks
```
